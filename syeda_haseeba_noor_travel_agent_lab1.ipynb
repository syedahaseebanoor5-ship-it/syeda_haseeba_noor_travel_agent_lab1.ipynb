```python
# Simple Travel Agent

print("=== Travel Agent ===")

destination = input("Enter your destination: ")
budget = int(input("Enter your budget in PKR: "))

if budget >= 10000:
    print("Travel Agent: A flight to", destination, "may be suitable for your budget.")
else:
    print("Travel Agent: Your budget is not enough for this option.")
    print("Please choose a cheaper flight.")
```
